## 09-06-2026 OOP AND FILE IO
OOP lets me bundle data and the functions that work on it into one object. A class is the blueprint, an object is the thing made from it.

Why classes: storing details for 23 students with separate variables does not scale. A class defines a Student once, then I stamp out as many as I want.

### Defining a Class

```python
# class - the blueprint. __init__ is the constructor, runs on creation.
# self is the object itself, so each object keeps its own data
class Student:
    def __init__(self, name, roll, marks):
        self.name = name
        self.roll = roll
        self.marks = marks

s1 = Student("Asha", 1, 88)
s2 = Student("Ravi", 2, 91)
print(s1.name, s1.marks)
print(s2.name, s2.marks)
```

### Methods

```python
# method - a function inside the class, uses self to read this object's data
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def result(self):
        return "Pass" if self.marks >= 40 else "Fail"

print(Student("Asha", 88).result())
print(Student("Meena", 32).result())
```

### __str__

```python
# __str__ - controls how the object prints (otherwise prints the memory address)
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def __str__(self):
        return f"Student({self.name}, {self.marks} marks)"

print(Student("Asha", 88))
```

### Encapsulation

```python
# encapsulation - protect data with __ (name mangling), expose safe methods
class SafeAccount:
    def __init__(self, balance):
        self.__balance = balance      # protected
    def withdraw(self, amount):
        if amount > self.__balance:
            return "Not enough balance"
        self.__balance -= amount
        return self.__balance
    def get_balance(self):
        return self.__balance

acc = SafeAccount(1000)
print(acc.withdraw(1500))
print(acc.withdraw(300))
```

### Inheritance

```python
# inheritance - child class reuses the parent. super() calls the parent constructor
class Person:
    def __init__(self, name):
        self.name = name
    def greet(self):
        return f"Hi, I am {self.name}"

class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks

s = Student("Asha", 88)
print(s.greet())          # inherited from Person
print(s.marks)
```

### Polymorphism and Overriding

```python
# polymorphism - same method name, different behaviour per object
class Dog:
    def sound(self): return "Bark"
class Cat:
    def sound(self): return "Meow"
for a in [Dog(), Cat()]:
    print(a.sound())
```

```python
# overriding - child redefines an inherited method
class Person:
    def greet(self): return "Hi, I am a person"
class Student(Person):
    def greet(self): return "Hi, I am a student"   # overrides parent
print(Student().greet())
```

### File I/O

```python
# open(path, mode). 'w' write/overwrite. with block auto-closes the file
with open("students.txt", "w") as f:
    f.write("Asha 88\n")
    f.write("Ravi 91\n")
print("written")
```

```python
# 'r' read the whole file
with open("students.txt", "r") as f:
    print(f.read())
```

```python
# 'a' append (keeps old content), then read line by line
with open("students.txt", "a") as f:
    f.write("Meena 32\n")
with open("students.txt", "r") as f:
    for line in f:
        print(line.strip())
```
