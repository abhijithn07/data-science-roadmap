## 09-06-2026 OOP AND FILE IO
Object Oriented Programming lets me bundle data and the functions that act on it into a single object. A class is a blueprint, and an object (instance) is a concrete thing built from it.

Why it matters: storing details for 23 students with separate variables (name1, marks1, name2, marks2 ...) does not scale and is impossible to manage. I define a Student class once, then create as many student objects as I need, each carrying its own data.

### Defining a Class
`class` names the blueprint. `__init__` is the constructor, a special method that runs automatically every time I create an object, and its job is to set up the starting data. `self` refers to the particular object being worked on; every method takes it as the first parameter so it knows whose data to use. I never pass self in by hand, Python does that. The values I pass as arguments get stored on the object with `self.x = x`, and each object keeps its own independent copy.

```python
# class is the blueprint, __init__ runs on creation, self is the object itself
class Student:
    def __init__(self, name, roll, marks):
        self.name = name
        self.roll = roll
        self.marks = marks

s1 = Student("Asha", 1, 88)
s2 = Student("Ravi", 2, 91)
print(s1.name, s1.marks)
print(s2.name, s2.marks)   # s2 has its own separate data
```

### Methods
A method is a function defined inside a class. Through `self` it can read and change that object's own data. Here `result()` decides pass or fail from this student's marks.

```python
# method - a function inside the class, uses self to read this object's data
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def result(self):
        return "Pass" if self.marks >= 40 else "Fail"

print(Student("Asha", 88).result())
print(Student("Meena", 32).result())
```

### __str__
By default printing an object shows an unhelpful memory address. Defining `__str__` lets me return a readable string, and Python calls it automatically whenever the object is printed or passed to str().

```python
# __str__ - controls how the object prints
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def __str__(self):
        return f"Student({self.name}, {self.marks} marks)"

print(Student("Asha", 88))
```

### Encapsulation
Encapsulation means keeping an object's data safe inside it and controlling how it is accessed. In Python the convention is a leading underscore `_x` for "internal", and two underscores `__x` which triggers name mangling so it cannot be reached by accident from outside. I then expose safe methods to read or change the value in valid ways. Below, the balance is protected and `withdraw` refuses to overdraw.

```python
# encapsulation - protect data with __ (name mangling), expose safe methods
class SafeAccount:
    def __init__(self, balance):
        self.__balance = balance          # protected
    def withdraw(self, amount):
        if amount > self.__balance:
            return "Not enough balance"
        self.__balance -= amount
        return self.__balance
    def get_balance(self):
        return self.__balance

acc = SafeAccount(1000)
print(acc.withdraw(1500))                 # blocked, too much
print(acc.withdraw(300))                  # allowed
print("balance:", acc.get_balance())
```

### Inheritance
Inheritance lets a new class reuse an existing one. The child automatically gets the parent's variables and methods, and can add its own. `class Child(Parent)` sets up the relationship, and `super().__init__(...)` calls the parent constructor so I do not repeat its setup. Here Student inherits from Person and adds marks.

```python
# inheritance - child reuses the parent, super() calls the parent constructor
class Person:
    def __init__(self, name):
        self.name = name
    def greet(self):
        return f"Hi, I am {self.name}"

class Student(Person):
    def __init__(self, name, marks):
        super().__init__(name)
        self.marks = marks

s = Student("Asha", 88)
print(s.greet())          # method inherited from Person
print(s.marks)
```

### Polymorphism and Overriding
Polymorphism means "many forms": the same method name behaves differently depending on the object, so I can loop over different objects and call the same method. Overriding is when a child class redefines a method it inherited, replacing the parent's version with its own.

```python
# polymorphism - same method name, different result per object
class Dog:
    def sound(self): return "Bark"
class Cat:
    def sound(self): return "Meow"
for a in [Dog(), Cat()]:
    print(a.sound())
```

```python
# overriding - the child redefines an inherited method
class Person:
    def greet(self): return "Hi, I am a person"
class Student(Person):
    def greet(self): return "Hi, I am a student"   # overrides the parent
print(Student().greet())
```

### File I/O
`open(path, mode)` opens a file. The mode picks the operation: `'r'` read (file must exist), `'w'` write (creates or overwrites), `'a'` append (adds to the end without erasing). The best practice is a `with` block, which closes the file automatically when the block ends, even if an error happens, so I never leak an open file handle. The `\n` is a newline so each record sits on its own line.

```python
# 'w' write/overwrite. The with block auto-closes the file
with open("students.txt", "w") as f:
    f.write("Asha 88\n")
    f.write("Ravi 91\n")
print("written")
```

```python
# 'r' read the whole file at once
with open("students.txt", "r") as f:
    print(f.read())
```

```python
# 'a' append keeps the old content, then read back line by line
with open("students.txt", "a") as f:
    f.write("Meena 32\n")
with open("students.txt", "r") as f:
    for line in f:
        print(line.strip())
```

### Recap
- class is the blueprint, object is the instance, `__init__` sets up data, `self` is the object
- methods act on self, `__str__` controls printing
- encapsulation protects data with `_` and `__` plus safe getters and setters
- inheritance reuses a parent (`super()`), polymorphism is one name many behaviours, overriding redefines an inherited method
- file i/o: `open(path, mode)` with modes r, w, a, always use a `with` block
